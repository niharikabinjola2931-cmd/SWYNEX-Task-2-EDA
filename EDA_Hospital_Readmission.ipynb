```python
readmission_by_admit_type = admissions.groupby("admit_type")["readmitted_30d"].mean() * 100

print("30D Readmission Rate by Admission Type")
print(readmission_by_admit_type.round(2))
```

```python
readmission_by_ward = admissions.groupby("ward_type")["readmitted_30d"].mean() * 100

print("30D Readmission Rate by Admission Type")
print(readmission_by_ward.round(2))
```

```python
admissions["loss_group"] = pd.cut(
    admissions["los_days"] , 
    bins = [ 0,3,7,14,float("inf")] , 
    labels = ["1-3days" , "4-7 days" , "8-14days" , "15+ days"]
)

print("30 Day Readmission Rate by length of Stay")
print(
    (admissions.groupby("loss_group" , observed = True)["readmitted_30d"].mean() * 100).round(2)
)
    
```

```python
readmission_by_charlson = admissions.groupby("charlson_index")["readmitted_30d"].mean() * 100

print("30 Day Readmission Rate by Charlson Index")
print(readmission_by_charlson.round(2))
```

```python
charlson_count = admissions["charlson_index"].value_counts().sort_index()

print("Number of Admissions by Charlson Index")
print(charlson_count.round(2))
```
